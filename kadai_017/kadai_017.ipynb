```python
class Human:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def check_adult(self):
    if self.age >= 20:
      print("大人である")
    else:
      print("大人でない")

human_list = [
      Human('alpha', 25),
      Human('beta', 19),
      Human('caro', 22),
      Human('david', 20),
      Human('eva', 27),
  ]
for user in human_list:
  user.check_adult()
```
