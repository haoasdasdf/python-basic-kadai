```python
from IPython.core.display_trap import DisplayTrap
class Human:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def check_adult(self):
      if self.age >= 20:
        print("大人である")
      else:
        print("大人でない")


data = [
      {'name': 'alpha', 'age': 25},
      {'name': 'beta', 'age': 19},
      {'name': 'caro', 'age': 22},
      {'name': 'david', 'age': 20},
      {'name': 'eva', 'age': 2}
  ]


for i in data:
  user = Human(i['name'], i['age'])
  user.check_adult()
```
