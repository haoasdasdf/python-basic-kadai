```python
class Human:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def check_adult(self):
    if self.age >= 20:
      print("大人である")
    else:
      print("大人でない")

human_list = [
      ['alpha', 25],
      ['beta', 19],
      ['caro', 22],
      ['david', 20],
      ['eva', 27],
  ]
for i in human_list:
  user = Human(i[0], i[1])
  user.check_adult()
```
